```python
# Rebecca Gabriella Blessing Sihite
# 065002500042
# Program Menentukan Rata-rata Nilai


def konversi_nilai(huruf):
    if huruf == "A":
        return 4
    elif huruf == "B":
        return 3
    elif huruf == "C":
        return 2
    elif huruf == "D":
        return 1
    elif huruf == "E":
        return 0
    else:
        return 0  # jika input tidak valid

# Program Perhitungan dan Input Nilai
jumlah_nilai = 0
banyak_nilai = 0

while True:
    huruf = input("Masukkan nilai: ").upper()
    if huruf == "":
        break
    nilai = konversi_nilai(huruf)
    if nilai == 0:
        print("Input tidak valid, coba lagi.")
    else:
        print("Nilai =", nilai)
        jumlah_nilai += nilai
        banyak_nilai += 1

if banyak_nilai > 0:
    rata_rata = jumlah_nilai / banyak_nilai
    print("Rata-rata nilainya adalah:", rata_rata)
else:
    print("Tidak ada nilai yang dimasukkan.")
```
