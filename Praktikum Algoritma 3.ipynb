```python
# Rebecca Gabriella Blessing Sihite
# 065002500042
# Program menentukan jenis segitiga berdasarkan panjang sisi

# Input tiga sisi segitiga
a = float(input("Masukkan sisi pertama: "))
b = float(input("Masukkan sisi kedua: "))
c = float(input("Masukkan sisi ketiga: "))

# Fungsi untuk cek apakah bisa membentuk segitiga
def jenis_segitiga(a, b, c):
    return a + b > c and a + c > b and b + c > a

if jenis_segitiga(a, b, c):
    # Cek jenis segitiga
    if a == b == c:
        print("Segitiga Sama Sisi")
    elif a == b or b == c or a == c:
        print("Segitiga Sama Kaki")
    elif (a**2 + b**2 == c**2) or (a**2 + c**2 == b**2) or (b**2 + c**2 == a**2):
        print("Segitiga Siku-Siku")
    else:
        print("Segitiga Sembarang")
else:
    print("Bukan segitiga (sisi tidak memenuhi syarat).")
```
