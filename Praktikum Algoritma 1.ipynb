```python
#Rebecca Gabriella Blessing Sihite
#065002500042

print("Hitung Luas Ruangan")
panjang_ruangan = int(input("Masukkan Panjang Ruangan: "))
lebar_ruangan = int(input("Masukkan Lebar Ruangan: "))
satuan = input("Masukkan Satuan (meter/inci): ")
if satuan == "meter":
    luas = panjang_ruangan * lebar_ruangan
    print(f"Luas ruangan dengan panjang {panjang_ruangan} dan lebar {lebar_ruangan} adalah {luas} {satuan} persegi.")
elif satuan == "inci":
    panjang_ruangan_inci = panjang_ruangan * 39.3701
    lebar_ruangan_inci = lebar_ruangan * 39.3701
    luas = panjang_ruangan_inci * lebar_ruangan_inci
    print(f"Luas ruangan dengan panjang {panjang_ruangan} meter ({panjang_ruangan_inci:.2f} inci) dan lebar {lebar_ruangan} meter ({lebar_ruangan_inci:.2f} inci) adalah {luas:.2f} {satuan} persegi.")
else:
    print("Satuan tidak valid. Harap masukkan 'meter' atau 'inci'.")
```
